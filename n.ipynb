```python
"""
EMPLOYEE ATTENDANCE SYSTEM
Python Project using Object-Oriented Programming (OOP)

Features:
1. Add Employee
2. View Employees
3. Mark Attendance
4. Calculate Monthly Attendance
5. Generate Attendance Report
"""

import csv
import os
from datetime import datetime

EMPLOYEE_FILE = "employees.csv"
ATTENDANCE_FILE = "attendance.csv"


# ---------------- EMPLOYEE CLASS ----------------

class Employee:

    def __init__(self, employee_id, name, department):
        self.employee_id = employee_id
        self.name = name
        self.department = department

    def to_row(self):
        return [
            self.employee_id,
            self.name,
            self.department
        ]


# ---------------- ATTENDANCE CLASS ----------------

class Attendance:

    def __init__(self, employee_id, date, status):
        self.employee_id = employee_id
        self.date = date
        self.status = status

    def to_row(self):
        return [
            self.employee_id,
            self.date,
            self.status
        ]


# ---------------- ATTENDANCE MANAGER CLASS ----------------

class AttendanceManager:

    def __init__(self):
        self.create_files()

    # Create CSV files
    def create_files(self):

        if not os.path.exists(EMPLOYEE_FILE):
            with open(EMPLOYEE_FILE, "w", newline="") as file:
                writer = csv.writer(file)
                writer.writerow([
                    "employee_id",
                    "name",
                    "department"
                ])

        if not os.path.exists(ATTENDANCE_FILE):
            with open(ATTENDANCE_FILE, "w", newline="") as file:
                writer = csv.writer(file)
                writer.writerow([
                    "employee_id",
                    "date",
                    "status"
                ])

    # Read employee records
    def read_employees(self):

        with open(EMPLOYEE_FILE, "r") as file:
            return list(csv.DictReader(file))

    # Read attendance records
    def read_attendance(self):

        with open(ATTENDANCE_FILE, "r") as file:
            return list(csv.DictReader(file))

    # Add employee
    def add_employee(self):

        employee_id = input("Enter Employee ID: ").strip()

        if not employee_id:
            print("Employee ID cannot be empty.")
            return

        employees = self.read_employees()

        if any(e["employee_id"] == employee_id for e in employees):
            print("Employee ID already exists.")
            return

        name = input("Enter Employee Name: ").strip()
        department = input("Enter Department: ").strip()

        if not name or not department:
            print("Name and Department are required.")
            return

        employee = Employee(
            employee_id,
            name,
            department
        )

        with open(EMPLOYEE_FILE, "a", newline="") as file:
            writer = csv.writer(file)
            writer.writerow(employee.to_row())

        print("Employee added successfully!")

    # View all employees
    def view_employees(self):

        employees = self.read_employees()

        if not employees:
            print("No employee records found.")
            return

        print("\n--------- EMPLOYEE DETAILS ---------")
        print("ID\tName\t\tDepartment")

        for e in employees:
            print(
                e["employee_id"],
                "\t",
                e["name"],
                "\t\t",
                e["department"]
            )

    # Mark attendance
    def mark_attendance(self):

        employee_id = input("Enter Employee ID: ").strip()

        employees = self.read_employees()

        if not any(e["employee_id"] == employee_id for e in employees):
            print("Employee not found!")
            return

        date = input(
            "Enter Date (YYYY-MM-DD) or press Enter for today: "
        ).strip()

        if not date:
            date = datetime.now().strftime("%Y-%m-%d")

        try:
            datetime.strptime(date, "%Y-%m-%d")
        except ValueError:
            print("Invalid date format!")
            return

        status = input(
            "Enter P for Present or A for Absent: "
        ).strip().upper()

        if status not in ["P", "A"]:
            print("Invalid attendance status!")
            return

        records = self.read_attendance()

        for record in records:
            if (
                record["employee_id"] == employee_id
                and record["date"] == date
            ):
                print("Attendance already recorded!")
                return

        attendance = Attendance(
            employee_id,
            date,
            status
        )

        with open(ATTENDANCE_FILE, "a", newline="") as file:
            writer = csv.writer(file)
            writer.writerow(attendance.to_row())

        print("Attendance marked successfully!")

    # Calculate monthly attendance
    def calculate_days(self):

        employee_id = input("Enter Employee ID: ").strip()
        month = input("Enter Month (YYYY-MM): ").strip()

        try:
            datetime.strptime(month, "%Y-%m")
        except ValueError:
            print("Invalid month format!")
            return

        employees = self.read_employees()

        employee = next(
            (
                e for e in employees
                if e["employee_id"] == employee_id
            ),
            None
        )

        if employee is None:
            print("Employee not found!")
            return

        records = [
            r for r in self.read_attendance()
            if r["employee_id"] == employee_id
            and r["date"].startswith(month)
        ]

        present_days = sum(
            1 for r in records if r["status"] == "P"
        )

        absent_days = sum(
            1 for r in records if r["status"] == "A"
        )

        total_days = present_days + absent_days

        print("\n====== MONTHLY ATTENDANCE REPORT ======")

        print("Employee ID:", employee_id)
        print("Employee Name:", employee["name"])
        print("Department:", employee["department"])
        print("Month:", month)

        print("Present Days:", present_days)
        print("Absent Days:", absent_days)
        print("Total Marked Days:", total_days)

        if total_days > 0:

            percentage = (
                present_days / total_days
            ) * 100

            print(
                "Attendance Percentage:",
                round(percentage, 2),
                "%"
            )

        else:
            print("No attendance records found.")

    # Main menu
    def run(self):

        while True:

            print("\n================================")
            print("   EMPLOYEE ATTENDANCE SYSTEM")
            print("================================")

            print("1. Add Employee")
            print("2. View Employees")
            print("3. Mark Attendance")
            print("4. Calculate Monthly Attendance")
            print("5. Exit")

            choice = input(
                "Enter your choice (1-5): "
            )

            if choice == "1":
                self.add_employee()

            elif choice == "2":
                self.view_employees()

            elif choice == "3":
                self.mark_attendance()

            elif choice == "4":
                self.calculate_days()

            elif choice == "5":
                print("Thank you for using Employee Attendance System!")
                break

            else:
                print("Invalid choice! Please try again.")


# ---------------- MAIN PROGRAM ----------------
121
if __name__ == "__main__":

    system = AttendanceManager()

    system.run()
```
